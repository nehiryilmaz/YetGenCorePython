# Problem 2

```python
# 1- (10,15,30,45,60) değerlerine sahip numpy dizisi oluşturunuz.
array1 = np.array([10,15,30,45,60])
print(array1)
# 2- (5-15) arasındaki sayılarla numpy dizisi oluşturunuz.
array2 = np.arange(5,16,1)
print(array2)
# 3- (50-100) arasında 5'er 5'er artarak numpy dizisi oluşturunuz.
array3 = np.arange(50,101,5)
print(array3)
# 4- 10 elemanlı sıfırlardan oluşan bir dizi oluşturunuz.
array4 = np.zeros(10)
print(array4)
# 5- 10 elemanlı birlerden oluşan bir dizi oluşturunuz.
array5 = np.ones(10)
print(array5)
# 6- (0-100) arasında eşit aralıklı 5 sayı üretin.
array6 = np.linspace(0,100,5)
print(array6)
# 7- (10-30) arasında rastgele 5 tane tamsayı üretin.
array7 = np.random.randint(10,31,5)
print(array7)
# 8- [-1 ile 1] arasında 10 adet sayı üretin.
array8 = np.random.uniform(-1,1,10)
print(array8)
# 9- (3x5) boyutlarında (10-50) arasında rastgele bir matris oluşturunuz.
matris = np.random.randint(10,51,(3,5))
print(matris)
# 10- Üretilen matrisin satır ve sütun sayıları toplamlarını hesaplayınız ?
satır_toplam = matris.sum(axis=1)
sütun_toplam = matris.sum(axis=0)
print(satır_toplam)
print(sütun_toplam)
# 11- Üretilen matrisin en büyük, en küçük ve ortalaması nedir ?
max = matris.max()
min = matris.min()
average = matris.mean()
print(max)
print(min)
print(average)
# 12- Üretilen matrisin en büyük değerinin indeksi kaçtır ?
max_indeks = np.unravel_index(matris.argmax(), matris.shape)
print(max_indeks)
# 13- (10-20) arasındaki sayıları içeren dizinin ilk 3 elemanını seçiniz.
array9 = np.arange(10,21,1)
print(array9[:3])
# 14- Üretilen dizinin elemanlarını tersten yazdırın.
print(array9[::-1])
# 15- Üretilen matrisin ilk satırını seçiniz.
print(matris[0])
# 16- Üretilen matrisin 2.satır 3.sütundaki elemanı hangisidir ?
print(matris[1,2])
# 17- Üretilen matrisin tüm satırlardaki ilk elemanı seçiniz.
print(matris[:,0])
# 18- Üretilen matrisin her bir elemanının karesini alınız.
print(matris**2)
# 19- Üretilen matris elemanlarının hangisi pozitif çift sayıdır ? Aralığı (-50,+50) arasında yapınız.
matris2 = np.random.randint(-50,51,(4,5))
print(matris2)
pozitif_cift_sayilar = matris2[(matris2 > 0) & (matris2 % 2 == 0)]
print(pozitif_cift_sayilar)
```
